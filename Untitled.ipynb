```python
import torch
from torch.nn import *

torch.set_default_tensor_type(torch.cuda.FloatTensor)


def flat(x):
    if isinstance(x, list):
        for x in x:
            yield from flat(x)
    else:
        yield x


class Linears(Linear):
    def forward(s, x):
        x = list(flat(x))
        if len(x) < 2:
            return super().forward(x[0])
        w = s.weight.split([x.size(-1) for x in x], 1)
        return sum([functional.linear(x, w) for x, w in zip(x[1:], w[1:])], functional.linear(x[0], w[0], s.bias))


class Nn(Sequential):
    def __init__(s, i, h, o=0):
        super().__init__(Linears(i, h), Hardswish(1), *[Linear(h, o)] if o else [])


class Notice(Nn):
    def __init__(s, q, k, h):
        super().__init__(q + k, h or k, 1)

    def forward(s, q, k):
        return super().forward([q[..., None, :] for q in flat(q)] + [k[..., None, :, :] for k in flat(k)])[..., 0].softmax(-1)


class Attention(Sequential):
    def __init__(s, q, k, v, h):
        super().__init__(Notice(q, k, h), Nn(v, h))

    def forward(s, q, k, v):
        return s[0](q, k) @ s[1](v)


class Bitention(Sequential):
    def __init__(s, q, k, h):
        super().__init__(Attention(1 + q, 1 + k, k, h), Attention(1 + k, 1 + q + h, q + h, h))

    def forward(s, q, k):
        Q = torch.linspace(0, 1, q.size(-2))[:, None]
        K = torch.linspace(0, 1, k.size(-2))[:, None]
        t = s[0]([Q, q], [K, k], k)
        return s[1]([K, k], [Q, q, t], [q, t])


class Kb(Sequential):
    def __init__(s, i, h, o, l):
        super().__init__(Bitention(i, h, h + h), Nn(h, h), Nn(h, h, o))
        s.l = l
        s.h = h

    def forward(s, x):
        a = torch.zeros(s.l, s.h)
        o = []
        for x in x:
            a, x = s[0](x, a).chunk(2, -1)
            o.append(x)
        return s[2](s[1](torch.stack(o)).mean(-2))
```

```python
from random import randint


def unpack(x, n, m=1 << torch.arange(16)):
    return (x[..., None] & m[:n]).bool().float()


def gets(a, k):
    s = {None, k}
    t = a.get(k)
    while t not in s:
        s.add(t)
        k = t
        t = a.get(t)
    return t or k


def qa(l, n):
    a = {}
    for _ in range(l):
        k = randint(0, n - 1)
        v = randint(0, n - 1)
        yield [k, v], gets(a, k)
        a[k] = v


def qas(l, b, n):
    a = [list(qa(l, n)) for _ in range(b)]
    return [torch.tensor([[a[j][k] for a in a] for j in range(l)]) for k in range(2)]
```

```python
n = 10
m = Kb(n, n, n, 2**n)
o = torch.optim.Adam(m.parameters())
O = torch.optim.lr_scheduler.OneCycleLR(o, 0.01, 10000, pct_start=0.1, div_factor=9, final_div_factor=99)
for t in range(O.total_steps // 100):
    for t in range(100):
        x, y = qas(2**n, 2, 2**n)
        x = m(unpack(x, n))
        y = unpack(y, n)
        t = functional.binary_cross_entropy_with_logits(x, y)
        t.backward()
        o.step()
        o.zero_grad()
        O.step()
    x = (x.signbit() != y).all(-1)
    print(t, x.sum() / x.numel())
```
