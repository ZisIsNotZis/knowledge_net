```python
import torch
from torch.nn import *

torch.set_default_tensor_type(torch.cuda.FloatTensor)


def flat(x):
    if isinstance(x, list):
        for x in x:
            yield from flat(x)
    else:
        yield x


class Linears(Linear):
    def forward(s, x):
        x = list(flat(x))
        if len(x) < 2:
            return super().forward(x[0])
        w = s.weight.split([x.size(-1) for x in x], 1)
        return sum([functional.linear(x, w) for x, w in zip(x[1:], w[1:])], functional.linear(x[0], w[0], s.bias))


class Nn(Sequential):
    def __new__(s, a, *_):
        if isinstance(a, int):
            return super().__new__(s)
        return Sequential(a)

    def __init__(s, i, h=0, o=0, inplace=1):
        super().__init__(Linears(i, h), Hardswish(inplace), *[Linear(h, o)] if o else [])


class Notice(Nn):
    def __init__(s, q, k, h):
        super().__init__(q + k + 1, h, 1)

    def forward(s, q, k):
        return super().forward([q, k, torch.linspace(0, 1, k.size(-2))[:, None]]).softmax(-2)


class Bitention(Module):
    def __init__(s, i, h, o):
        super().__init__()
        s.read = Notice(i, h, h)
        s.readNn = Nn(h, h)
        s.write = Notice(h, h, h)
        s.nn = Nn(i + h, h + h)
        s.out = Linear(h, o)

    def forward(s, i, h):
        o, i = s.nn([i, s.read(i, h)[..., None, :, 0] @ s.readNn(h)]).chunk(2, -1)
        t = s.write(i, h)
        return s.out(o), t * i + (1 - t) * h


class Kb(Bitention):
    def __init__(s, i, h, o, l):
        super().__init__(i, h, o)
        s.l = h
        s.h = h

    def forward(s, x):
        h = torch.empty(s.l, s.h)
        o = []
        for x in x:
            x, h = super().forward(x, h)
            o.append(x)
        return torch.stack(o)
```

```python
class Vec2Seq(Nn):
    def __new__(s, a, *_):
        if isinstance(a, int):
            return super().__new__(s)
        return Sequential(a)

    def __init__(s, i, h, o):
        super().__init__(i, h, h + 1 + o, inplace=0)
        s.h = h
        s.o = o

    def forward(s, x, l=4):
        x, p, o = super().forward(x).split([s.h, 1, s.o], -1)
        p = torch.tensor([0.9])
        P = p.sigmoid()
        O = [o]
        for i in range(2, l + 1):
            x, p, o = s[1:](x).split([s.h, 1, s.o], -1)
            p = torch.tensor([0.9])
            P *= p.sigmoid() if i < l else 1 - p.sigmoid()
            O.append(o)
        return P, torch.stack(O).softmax(-1)


class Dense(Module):
    def __init__(s, a):
        super().__init__()
        s.m = {a: i for i, a in enumerate(a)}
        s.M = {i: a for i, a in enumerate(a)}

    def forward(s, x, box=1):
        if isinstance(x, str):
            x = [s.m[x] for x in x]
        else:
            x = [s(x, 0) for x in x]
        return torch.tensor(x) if box else x

    def rev(s, x):
        return "".join(s.M[x] for x in x)


class GRUs(GRU):
    def __init__(s, *a, **k):
        return super().__init__(*a, **k, batch_first=True)

    def forward(s, x):
        return super().forward(x.view(-1, x.size(-2), x.size(-1)))[1].view(*x.shape[:-2], x.size(-1))


class Seq2Vec(Sequential):
    def __new__(s, a, *_):
        if isinstance(a, int):
            return super().__new__(s)
        return Sequential(a)

    def __init__(s, k, a):
        super().__init__(Dense(a), Embedding(len(a), k), GRUs(k, k))
```

```python
from random import choice


def qa(l, K="abcd", V="0123"):
    a = {}
    for _ in range(l):
        k = choice(K)
        v = choice(V)
        o = a.get(k)
        if not o:
            a[k] = v
            o = "ok"
        elif o != v:
            o = k + "=" + o
        else:
            o = "yes"
        yield k + "=" + v, o


def qas(l, b):
    a = [list(qa(l)) for _ in range(b)]
    return [[[a[j][k] for a in a] for j in range(l)] for k in range(2)]
```
