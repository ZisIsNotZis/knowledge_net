```python
import torch
from torch.nn import *

ACT = Hardswish(1)
torch.set_default_tensor_type(torch.cuda.FloatTensor)


class F(Module):
    def __init__(s, f, a=None):
        super().__init__()
        s.f = f
        s.a = a

    def forward(s, x):
        return s.f(x) if s.a is None else s.f(x, s.a)


class Cat(tuple):
    def __new__(s, *a):
        return super().__new__(s, (a for a in a for a in (a if isinstance(a, Cat) else (a,))))

    def __mul__(s, a):
        return Cat(*(s * a for s in s))


class Linear(torch.nn.Linear):
    def forward(s, x):
        if isinstance(x, Cat):
            w = s.weight.split([x.size(-1) for x in x], 1)
            return sum([functional.linear(x, w) for x, w in zip(x, w[:-1])], functional.linear(x[-1], w[-1], s.bias))
        return super().forward(x)


class Nn(Sequential):
    def __init__(s, i, h, o=0):
        super().__init__(Linear(i, h), ACT, *(Linear(h, o),) if o else ())
```

```python
class Notice(Nn):
    def __init__(s, q, k, h):
        super().__init__(q + k + 1, h, 1)

    def forward(s, q, k):
        return super().forward(Cat(q, k, torch.linspace(0, 1, k.size(-2))[:, None])).softmax(-2)


class KbCell(Module):
    def __init__(s, i, h):
        super().__init__()
        s.read = Notice(i, h, h)
        s.compute = Nn(i + h, h)
        s.write = Notice(h, h, h)
        s.learn = Nn(h + h, h)
        s.h = h

    def forward(s, x, a):
        x = s.compute(Cat(x, s.read(x, a)[..., None, :, 0] @ a))
        return x, s.learn(Cat(x * s.write(x, a), a))


class Kb(KbCell):
    def forward(s, x, l):
        a = torch.empty(l, s.h)
        X = []
        for x in x:
            x, a = super().forward(x, a)
            X.append(x)
        return torch.stack(X)
```

```python
class Red(Sequential):
    def forward(s, x):
        while x.size(-1) > 1:
            x = functional.avg_pool1d(x + super().forward(x), 2)
        return x


class Seq2Vec(Sequential):
    def __init__(s, i, h):
        super().__init__(Conv1d(i, h, 1), ACT, Red(Conv1d(h, h, 3, padding=1, bias=0), BatchNorm1d(h, track_running_stats=0), ACT))


class Exp(Sequential):
    def __init__(s, l, *a):
        super().__init__(*a)
        s.l = l

    def forward(s, x):
        while x.size(-1) < s.l:
            x = x.repeat_interleave(2, -1)
            x = x + super().forward(x)
        return x


class Vec2Seq(Sequential):
    def __init__(s, i, h, l, act=1):
        super().__init__(Conv1d(i, h, 1), *(ACT,) if act else (), Exp(l, ConvTranspose1d(h, h, 3, padding=1, bias=0), BatchNorm1d(h, track_running_stats=0), ACT))


class Seq2Seq(Sequential):
    def __init__(s, i, h, o, l, H=0):
        super().__init__(Seq2Vec(i, h), *(Conv1d(h, H, 1), ACT) if H else (), Vec2Seq(H or h, h, l, 0), Conv1d(h, h, 1), ACT, Conv1d(h, o, 1))
```

```python
from itertools import combinations
import numpy as np


def seqlos_(y, t):
    if isinstance(t[0], list):
        return sum(seqlos_(y, t) for y, t in zip(y, t)) / len(t)
    return y[torch.tensor(list(combinations(range(len(y)), len(t)))), t].prod(1).sum().log()


def seqlos(y, t):
    return -seqlos_(y.softmax(-1), t)


class Map(dict):
    def __init__(s, a):
        super().__init__((a, i) for i, a in enumerate(a))
        s.a = a

    def __call__(s, x, l=0):
        if isinstance(x, str):
            x = [s[x] for x in x]
            return x + [0] * (l - len(x)) if l else x
        return [s(x, l) for x in x]

    def inv(s, x):
        return "".join(s.a[x] for x in x)


def strlen(a):
    t, l = a.min(-1)
    l[t.bool()] = a.size(-1)
    return l


def seqlos(y, t):
    y = y.log_softmax(2)
    l = strlen(t)
    o = 0
    for i in torch.unique(l):
        m = (l == i).nonzero()
        c = torch.tensor(list(combinations(range(y.size(1)), i)))
        T = torch.zeros(len(m), len(c), y.size(1), dtype=torch.long)
        T[torch.arange(len(m))[:, None, None], torch.arange(len(c))[:, None], c] = t[m, :i]
        o += y[m[:, None], torch.arange(y.size(1)), T].sum(2).logsumexp(1).sum()
    return -o / len(l)


class Map(np.ndarray):
    def __new__(s, a):
        return np.array(tuple(a), "B").view(s)

    def __init__(s, a):
        s.m = np.zeros(127, "B")
        s.m[s] = range(len(s))

    def __call__(s, x):
        return s.m[x[:, None].view("B")]

    def inv(s, x):
        return s[x.cpu()].view(f"S{len(x)}")[0]


def strcat(*a):
    while len(a) > 1:
        a = tuple(np.char.add(*a) for a in zip(a[::2], a[1::2])) + a[len(a) // 2 * 2 :]
    return a[0]


def bieval(a, o, b):
    return np.stack((a & b, a | b, a ^ b, a + b, a - b))[o, np.arange(len(a))]


def dat_(a, A=None, O=np.array((b"&", b"|", b"^", b"+", b"-"))):
    A = a.astype("S4") if A is None else strcat(b"(", A, b")")
    o = np.random.randint(0, 5, len(a))
    b = np.random.randint(0, 10000, len(a))
    B = b.astype("S4")
    w = np.random.randint(0, 2, len(a))
    return np.where(w, bieval(a, o, b), bieval(b, o, a)), np.where(w, strcat(A, O[o], B), strcat(B, O[o], A))


def dat(n, k=0):
    a, A = dat_(np.random.randint(0, 10000, n))
    for k in range(k):
        a, A = dat_(a, A)
    return A, np.char.rjust((a % 10000).astype("S4"), 4, b"0")
```

```python
ma = Map(b"_1234567890&|^+-")
emb = Embedding(len(ma), 64)
m = Seq2Seq(64, 64, len(ma), 4)
o = torch.optim.Adam((*emb.parameters(), *m.parameters()))
O = torch.optim.lr_scheduler.OneCycleLR(o, 0.03, 100000, pct_start=0.1, div_factor=9, final_div_factor=99)
for x in range(O.total_steps // 1000):
    for x in range(1000):
        x, t = dat(1024)
        y = m(emb(torch.tensor(ma(x), dtype=torch.long)).transpose(1, 2)).transpose(1, 2)
        l = functional.cross_entropy(y.transpose(1, 2), torch.tensor(ma(t), dtype=torch.long))
        l.backward()
        o.step()
        o.zero_grad()
        O.step()
    print(l, x[0], t[0], ma.inv(y[0].argmax(-1)))
```
