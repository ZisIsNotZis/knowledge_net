```python
import torch
import numpy as np
from torch.nn import *

torch.set_default_tensor_type(torch.cuda.FloatTensor)


class F(Module):
    def __init__(s, f, a=None):
        super().__init__()
        s.f = f
        s.a = a

    def forward(s, x):
        return s.f(x) if s.a is None else s.f(x, s.a)


class Cat(tuple):
    def __new__(s, *a):
        return super().__new__(s, (a for a in a for a in (a if isinstance(a, Cat) else (a,))))

    def __mul__(s, a):
        return Cat(*(s * a for s in s))


class Linear(torch.nn.Linear):
    def forward(s, x):
        if isinstance(x, Cat):
            w = s.weight[:, : sum(x.size(-1) for x in x)].split([x.size(-1) for x in x], 1)
            return sum([functional.linear(x, w) for x, w in zip(x, w[:-1])], functional.linear(x[-1], w[-1], s.bias))
        return super().forward(x)


class Nn(Sequential):
    def __init__(s, i, h, o=0):
        super().__init__(Linear(i, h), ReLU(1), *(Linear(h, o),) if o else ())


def cross_entropy(y, t):
    return functional.cross_entropy(y.permute(0, y.ndim - 1, *range(1, y.ndim - 1)), t)


class Notice(Nn):
    def __init__(s, q, k, h, head=1):
        super().__init__(q + k + 1, head * h, head)

    def forward(s, q, k):
        return super().forward(Cat(q, k, torch.linspace(0, 1, k.size(-2))[:, None])).softmax(-2)


class KbCell(Module):
    def __init__(s, i, h, o, headin=1, headout=1):
        super().__init__()
        s.read = Notice(i, h, h, headin)
        s.compute = Nn(i + headin * h, o)
        s.write = Notice(o, h, h, headout)
        s.save = Nn(o + (0 if headout < 2 else headout) + h, h)
        s.h = h
        s.headout = headout

    def forward(s, x, a):
        x = s.compute(Cat(x, (s.read(x, a).transpose(-2, -1) @ a).flatten(-2)[..., None, :]))
        return x, s.save(Cat(x * s.write(x, a), a) if s.headout < 2 else Cat(x, s.write(x, a), a))


class Kb(KbCell):
    def __init__(s, i, h, o, l, headin=1, headout=1):
        super().__init__(i, h, o, headin, headout)
        s.l = l

    def forward(s, x):
        a = torch.zeros(s.l, s.h)
        X = []
        for x in x:
            x, a = super().forward(x, a)
            X.append(x)
        return torch.stack(X)


class Akb(Module):
    def __init__(s, i, h, o, head=1):
        super().__init__()
        s.read = Notice(i, h, h, head)
        s.compute = Nn(i + head * h, h + o)
        s.h = h

    def forward(s, x):
        A = None
        X = []
        for x in x:
            x = s.compute(Cat(*[x] if A is None else (x, (s.read(x[..., None, :], A).transpose(-2, -1) @ A).flatten(-2))))
            x, a = x.split((x.size(-1) - s.h, s.h), -1)
            a = a[..., None, :]
            A = a if A is None else torch.cat((A, a), -2)
            X.append(x)
        return torch.stack(X)
```

```python
class ConvSpace(Sequential):
    def forward(s, x):
        return super().forward(x.flatten(end_dim=-3).transpose(1, 2)).transpose(1, 2).unflatten(0, x.shape[:-2])


class Red(Sequential):
    def forward(s, x):
        while x.size(-1) > 1:
            x = super().forward(x)
        return x


class Seq2Vec(Sequential):
    def __init__(s, i, h):
        super().__init__(Linear(i, h), ReLU(1), ConvSpace(Red(Conv1d(h, h, 3, padding=1, bias=0), BatchNorm1d(h, track_running_stats=0), ReLU(1))))


class Exp(Sequential):
    def __init__(s, l, *a):
        super().__init__(*a)
        s.l = l

    def forward(s, x):
        while x.size(-1) < s.l:
            x = x.repeat_interleave(2, -1)
            x = x + super().forward(x)
        return x


class Vec2Seq(Sequential):
    def __init__(s, i, h, l, act=1):
        super().__init__(
            Linear(i, h), *(ReLU(1),) if act else (), ConvSpace(Exp(l, ConvTranspose1d(h, h, 3, padding=1, bias=0), BatchNorm1d(h, track_running_stats=0), ReLU(1)))
        )


class Seq2Seq(Sequential):
    def __init__(s, i, h, o, l, H=0):
        super().__init__(Seq2Vec(i, h), *(Linear(h, H), ReLU(1)) if H else (), Vec2Seq(H or h, h, l, 0), Linear(h, h), ReLU(1), Linear(h, o))
```

```python
a = Dat(128, 20, 16, 256)
k = a.l
m = Sequential(
    Embedding(len(a), k),
    F(lambda x: x.flatten(1, 2).permute(1, 2, 0)),
    Red(
        Res(
            Conv1d(k, k, 3, padding=1),
            ReLU(1),
            Conv1d(k, k, 1),
        ),
        AvgPool1d(2),
    ),
    F(lambda x: x[..., 0].unflatten(0, (a.L, a.n))),
    Akb(k, k, 512),
    Linear(512, len(a)),
)
o = torch.optim.Adam(m.parameters())
O = torch.optim.lr_scheduler.OneCycleLR(o, 0.02, 30000, pct_start=0.01, div_factor=9, final_div_factor=999)
log = Log()
for i in range(O.total_steps // 300):
    for i in range(300):
        x, t = a()
        y = m(x)
        l = functional.cross_entropy(y.transpose(1, 2), t)
        l.backward()
        o.step()
        o.zero_grad()
        O.step()
    log(l.detach().cpu(), (y.argmax(-1) == t).sum(-1).cpu() / t.size(-1))
```
