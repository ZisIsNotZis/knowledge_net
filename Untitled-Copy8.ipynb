```python
import torch
import numpy as np
from torch.nn import *

torch.set_default_tensor_type(torch.cuda.FloatTensor)


class F(Module):
    def __init__(s, f, a=None):
        super().__init__()
        s.f = f
        s.a = a

    def forward(s, x):
        return s.f(x) if s.a is None else s.f(x, s.a)


class Cat(tuple):
    def __new__(s, *a):
        return super().__new__(s, (a for a in a for a in (a if isinstance(a, Cat) else (a,))))

    def __mul__(s, a):
        return Cat(*(s * a for s in s))


class Linear(torch.nn.Linear):
    def forward(s, x):
        if isinstance(x, Cat):
            return sum(
                functional.linear(x, w, None if i else s.bias)
                for i, (x, w) in enumerate(zip(x, s.weight[:, : sum(x.size(-1) for x in x)].split([x.size(-1) for x in x], 1)))
            )
        return super().forward(x)


class BatchNormFlat(BatchNorm1d):
    def forward(s, x):
        return super().forward(x.flatten(end_dim=-2)).unflatten(0, x.shape[:-1]) if x.ndim > 2 else super().forward(x)


class Nn(Sequential):
    def __init__(s, i, h, o=0):
        super().__init__(Linear(i, h, 0), BatchNormFlat(h, track_running_stats=0), Hardswish(1), *(Linear(h, o),) if o else ())


class Notice(Nn):
    def __init__(s, q, k, h, head=1):
        super().__init__(q + k + 1, head * h, head)

    def forward(s, q, k):
        return super().forward(Cat(q, k, torch.linspace(0, 1, k.size(-2))[:, None])).softmax(-2)


class KbCell(Module):
    def __init__(s, i, h, o, headin=1, headout=1):
        super().__init__()
        s.read = Notice(i, h, h, headin)
        s.compute = Nn(i + headin * h, o)
        s.write = Notice(o, h, h, headout)
        s.save = Nn(o + (0 if headout < 2 else headout) + h, h)
        s.h = h
        s.headout = headout

    def forward(s, x, a):
        x = s.compute(Cat(x, (s.read(x, a).transpose(-2, -1) @ a).flatten(-2)[..., None, :]))
        return x, s.save(Cat(x * s.write(x, a), a) if s.headout < 2 else Cat(x, s.write(x, a), a))


class Kb(KbCell):
    def __init__(s, i, h, o, l, headin=1, headout=1):
        super().__init__(i, h, o, headin, headout)
        s.l = l

    def forward(s, x):
        a = torch.zeros(s.l, s.h)
        X = []
        for x in x:
            x, a = super().forward(x, a)
            X.append(x)
        return torch.stack(X)


class Akb(Module):
    def __init__(s, i, h, o, head=1):
        super().__init__()
        s.read = Notice(i, h, h, head)
        s.compute = Nn(i + head * h, h + o)
        s.h = h

    def forward(s, x):
        A = None
        X = []
        for x in x:
            x = s.compute(Cat(*[x] if A is None else (x, (s.read(x[..., None, :], A).transpose(-2, -1) @ A).flatten(-2))))
            x, a = x.split((x.size(-1) - s.h, s.h), -1)
            a = a[..., None, :]
            A = a if A is None else torch.cat((A, a), -2)
            X.append(x)
        return torch.stack(X)


class Red(Sequential):
    def forward(s, x):
        while x.size(-1) > 1:
            x = super().forward(x)
        return x


class Res(Sequential):
    def forward(s, x):
        return super().forward(x) + x
```

```python
import matplotlib.pyplot as plt
from IPython.display import clear_output


def inews(a="../nlpdat/inews/dev.txt"):
    a = np.loadtxt(a, "U", None, "_!_", usecols=(2, 3), encoding="utf-8").view("I").view("2H")[..., 0]
    a[:, -1] = ord("。")
    return a[(0x4E00 <= a) & (a <= 0x9FA5)]


def roll(a, n):
    return np.lib.stride_tricks.as_strided(a[:0].view(f"V{n}" if n != 2 else "H"), [len(a) - n + 1], [1])


def entropy(a):
    if a.dtype in ("B", "H"):
        m = np.zeros(np.iinfo(a.dtype).max + 1, "I")
        np.add.at(m, a, 1)
        m = m[m.astype("?")]
    else:
        m = np.unique(a, return_counts=1)[1]
    m += 1
    c = np.power(256.0, a.itemsize)
    M = len(a) + c
    m = np.divide(m, M)
    return m @ np.log2(m) - (c - len(m)) * np.log2(M) / M


def entropy2(a):
    if a.dtype in ("B", "H"):
        m = np.zeros(np.iinfo(a.dtype).max + 1, "I")
        np.add.at(m, a, 1)
        m = m[m.astype("?")]
    else:
        m = np.unique(a, return_counts=1)[1]
    c = np.power(256.0, a.itemsize) - len(m)
    M = c + len(a)
    m = np.divide(m, M)
    return m @ np.log2(m) - c * np.log2(M) / M


class Dat:
    def __init__(s, a, n, L, l, r):
        t = np.ones(65536, "I")
        np.add.at(t, a, 1)
        s.M = t.argpartition(-257)[-256:]
        t = t[s.M]
        t = t / t.sum()
        print(t @ np.log2(t))
        s.m = np.full(65536, -1, "B")
        s.m[s.M] = np.arange(len(s.M), dtype="B")
        a = s.m[a]
        s.a = a[a >= 0]
        s.x = torch.tensor(s.a)
        s.n = n
        s.L = L
        s.l = l
        s.r = r

    def __call__(s, n=0, L=0, l=0, r=0):
        n = n or s.n
        L = L or s.L
        l = l or s.l
        r = r or s.r
        x, t = s.x[torch.randint(len(s.x) - r + 1, (n, 1)) + torch.randint(r - l, (L, n, 1)) + torch.arange(l + 1)].long().split((l, 1), 2)
        return x, t[..., 0]

    def __len__(s):
        return len(s.M)


class DatNoL(Dat):
    def __call__(s):
        x, t = s.x[torch.randint(len(s.x) - s.l, (s.n,)) + torch.arange(s.l + 1)[:, None]].long().split((s.l, 1), 0)
        return x, t[0]


class DatMock:
    def __init__(s, n, L, l, r, c=256):
        s.n = n
        s.L = L
        s.l = l
        s.r = r
        s.c = c

    def __call__(s):
        a = torch.randint(s.c, (s.n, s.r))
        x, t = a[torch.arange(s.n), torch.randint(s.r - s.l, (s.L, s.n)) + torch.arange(s.l + 1)[:, None, None]].split((s.l, 1), 0)
        return x, t[0]


class Log:
    a = None

    def __call__(s, *x):
        x = [x[None] for x in x]
        s.a = x if s.a is None else [np.concatenate((s, x)) for s, x in zip(s.a, x)]
        for i, x in enumerate(s.a):
            plt.subplot(len(s.a), 1, i + 1).plot(x)
        clear_output(1)
        plt.show()
```

```python
a = Dat(256, 16, 16, 64)
k = a.l
m = Sequential(
    Embedding(len(a), k),
    F(lambda x: x.flatten(end_dim=1).transpose(1, 2)),
    Red(
        Res(
            Conv1d(k, k, 3, padding=1, bias=0),
            BatchNorm1d(k, track_running_stats=0),
            Hardswish(1),
            Conv1d(k, k, 3, padding=1),
        ),
        MaxPool1d(2),
    ),
    F(lambda x: x[..., 0].unflatten(0, (a.L, a.n))),
    Akb(k, k, 512),
    Linear(512, len(a)),
)
o = torch.optim.Adam(m.parameters())
O = torch.optim.lr_scheduler.OneCycleLR(o, 0.05, 10000, pct_start=0.01, div_factor=9, final_div_factor=999)
log = Log()
for i in range(O.total_steps // 100):
    for i in range(100):
        x, t = a()
        y = m(x)
        l = functional.cross_entropy(y.transpose(1, 2), t)
        l.backward()
        o.step()
        o.zero_grad()
        O.step()
    log(l.detach().cpu(), (y.argmax(-1) == t).sum(-1).cpu() / t.size(-1))
```

```python
class MyAkb(Akb):
    def __init__(s, i, h, o, head=1, w=1):
        Module.__init__(s)
        s.read = Notice(i, h, (i + h) * w, head)
        s.compute = Nn(i + h * head, max(i + h * head, h + o) * w, h + o)
        s.h = h


a = Dat(np.concatenate((inews(), inews("../nlpdat/inews/test.txt"), inews("../nlpdat/inews/train.txt"))), 8, 8, 16, 64)
k = 8
K = k * a.l
w = 8
m = Sequential(
    Embedding(256, k),
    Nn(k, K * w, K),
    F(lambda x: x.flatten(end_dim=1).transpose(1, 2)),
    Red(
        Res(
            Conv1d(K, K * w, 3, padding=1, bias=0),
            BatchNorm1d(K * w, track_running_stats=0),
            Hardswish(1),
            Conv1d(K * w, K, 1),
        ),
        MaxPool1d(2),
    ),
    F(lambda x: x[..., 0].unflatten(0, (a.L, len(x) // a.L))),
    MyAkb(K, K, 256, w=w),
)
o = torch.optim.Adam(m.parameters())
O = torch.optim.lr_scheduler.OneCycleLR(o, 0.01, 50000, pct_start=0.01, div_factor=9, final_div_factor=99)
log = Log()
for i in range(O.total_steps // 500):
    L = 0
    A = 0
    for i in range(5):
        for i in range(100):
            x, t = a()
            y = m(x)
            l = functional.cross_entropy(y.transpose(1, 2), t)
            l.backward()
            o.step()
            o.zero_grad()
            O.step()
        L += l.detach()
        A += (y.argmax(-1) == t).float().mean(-1)
    log(L.cpu() / 5, (A / 5).cpu())
```
