```python
import torch
from torch.nn import *

torch.set_default_tensor_type(torch.cuda.FloatTensor)


def flat(x):
    if isinstance(x, list):
        for x in x:
            yield from flat(x)
    else:
        yield x


class Linears(Linear):
    def forward(s, x):
        x = list(flat(x))
        if len(x) < 2:
            return super().forward(x[0])
        w = s.weight.split([x.size(-1) for x in x], 1)
        return sum([functional.linear(x, w) for x, w in zip(x[1:], w[1:])], functional.linear(x[0], w[0], s.bias))


class Nn(Sequential):
    def __init__(s, i, h, o=0):
        super().__init__(Linears(i, h), Hardswish(1), *[Linear(h, o)] if o else [])


class Notice(Nn):
    def __init__(s, q, k, h):
        super().__init__(q + k + 1, h, 1)

    def forward(s, q, k):
        return super().forward([q, k, torch.linspace(0, 1, k.size(-2))[:, None]]).softmax(-2)


class Bitention(Module):
    def __init__(s, i, h, o):
        super().__init__()
        s.read = Notice(i, h, h)
        s.readNn = Nn(h, h)
        s.write = Notice(h, h, h)
        s.nn = Nn(i + h, h + h)
        s.out = Linear(h, o)

    def forward(s, i, h):
        o, i = s.nn([i, s.read(i, h)[..., None, :, 0] @ s.readNn(h)]).chunk(2, -1)
        t = s.write(i, h)
        return s.out(o), t * i + (1 - t) * h


class Kb(Bitention):
    def __init__(s, i, h, o, l):
        super().__init__(i, h, o)
        s.l = h
        s.h = h

    def forward(s, x):
        h = torch.empty(s.l, s.h)
        o = []
        for x in x:
            x, h = super().forward(x, h)
            o.append(x)
        return torch.stack(o)
```

```python
m = Kb(1, 4, 3, 4)
o = torch.optim.Adam(m.parameters())
O = torch.optim.lr_scheduler.OneCycleLR(o, 0.01, 10000, pct_start=0.1, div_factor=9, final_div_factor=99)
for t in range(O.total_steps // 100):
    for t in range(100):
        t = torch.randn(64, 4, 1)
        t = functional.smooth_l1_loss(m(t), torch.cat([t.cummin(0)[0], t.cummax(0)[0], t.cumsum(0) / torch.arange(1.0, 1 + len(t))[:, None, None]], 2))
        t.backward()
        o.step()
        o.zero_grad()
        O.step()
    print(t)
```
