```python
import torch
import numpy as np
from torch.nn import *

torch.set_default_tensor_type(torch.cuda.FloatTensor)


class F(Module):
    def __init__(s, f, a=None):
        super().__init__()
        s.f = f
        s.a = a

    def forward(s, x):
        return s.f(x) if s.a is None else s.f(x, s.a)


class Cat(tuple):
    def __new__(s, *a):
        return super().__new__(s, (a for a in a for a in (a if isinstance(a, Cat) else (a,))))

    def __mul__(s, a):
        return Cat(*(s * a for s in s))


class Linear(torch.nn.Linear):
    def forward(s, x):
        if isinstance(x, Cat):
            w = s.weight.split([x.size(-1) for x in x], 1)
            return sum([functional.linear(x, w) for x, w in zip(x, w[:-1])], functional.linear(x[-1], w[-1], s.bias))
        return super().forward(x)


class Nn(Sequential):
    def __init__(s, i, h, o=0):
        super().__init__(Linear(i, h), ReLU(1), *(Linear(h, o),) if o else ())


def cross_entropy(y, t):
    return functional.cross_entropy(y.permute(0, y.ndim - 1, *range(1, y.ndim - 1)), t)


class Notice(Nn):
    def __init__(s, q, k, h):
        super().__init__(q + k + 1, h, 1)

    def forward(s, q, k):
        return super().forward(Cat(q, k, torch.linspace(0, 1, k.size(-2))[:, None])).softmax(-2)


class KbCell(Module):
    def __init__(s, i, h):
        super().__init__()
        s.read = Notice(i, h, h)
        s.compute = Nn(i + h, h + h)
        s.write = Notice(h, h, h)
        s.h = h

    def forward(s, x, a):
        o, x = s.compute(Cat(x, s.read(x, a)[..., None, :, 0] @ a)).chunk(2, -1)
        w = s.write(x, a)
        return o, w * x + (1 - w) * a


class Kb(KbCell):
    def __init__(s, i, h, l):
        super().__init__(i, h)
        s.l = l

    def forward(s, x):
        a = torch.zeros(s.l, s.h)
        X = []
        for x in x:
            x, a = super().forward(x, a)
            X.append(x)
        return torch.stack(X)
```

```python
class ConvSpace(Sequential):
    def forward(s, x):
        return super().forward(x.flatten(end_dim=-3).transpose(1, 2)).transpose(1, 2).unflatten(0, x.shape[:-2])


class Red(Sequential):
    def forward(s, x):
        while x.size(-1) > 1:
            x = functional.avg_pool1d(x + super().forward(x), 2)
        return x


class Seq2Vec(Sequential):
    def __init__(s, i, h):
        super().__init__(Linear(i, h), ReLU(1), ConvSpace(Red(Conv1d(h, h, 3, padding=1, bias=0), BatchNorm1d(h, track_running_stats=0), ReLU(1))))


class Exp(Sequential):
    def __init__(s, l, *a):
        super().__init__(*a)
        s.l = l

    def forward(s, x):
        while x.size(-1) < s.l:
            x = x.repeat_interleave(2, -1)
            x = x + super().forward(x)
        return x


class Vec2Seq(Sequential):
    def __init__(s, i, h, l, act=1):
        super().__init__(Linear(i, h), *(ReLU(1),) if act else (), ConvSpace(Exp(l, ConvTranspose1d(h, h, 3, padding=1, bias=0), BatchNorm1d(h, track_running_stats=0), ReLU(1))))


class Seq2Seq(Sequential):
    def __init__(s, i, h, o, l, H=0):
        super().__init__(Seq2Vec(i, h), *(Linear(h, H), ReLU(1)) if H else (), Vec2Seq(H or h, h, l, 0), Linear(h, h), ReLU(1), Linear(h, o))
```

```python
from itertools import combinations


def strlen(a):
    t, l = a.min(-1)
    l[t.bool()] = a.size(-1)
    return l


def strcat(*a):
    while len(a) > 1:
        a = tuple(np.char.add(*a) for a in zip(a[::2], a[1::2])) + a[len(a) // 2 * 2 :]
    return a[0]


def seqlos(y, t):
    y = y.log_softmax(2)
    l = strlen(t)
    o = 0
    for i in torch.unique(l):
        m = (l == i).nonzero()
        c = torch.tensor(list(combinations(range(y.size(1)), i)))
        T = torch.zeros(len(m), len(c), y.size(1), dtype=torch.long)
        T[torch.arange(len(m))[:, None, None], torch.arange(len(c))[:, None], c] = t[m, :i]
        o += y[m[:, None], torch.arange(y.size(1)), T].sum(2).logsumexp(1).sum()
    return -o / len(l)


class Map(Module):
    def __init__(s, a):
        super().__init__()
        s.a = np.array(tuple(a), "B")
        s.m = np.zeros(127, "B")
        s.m[s.a] = range(len(s.a))

    def __call__(s, x):
        return torch.tensor(s.m[x[..., None].view("B")], dtype=torch.long)

    def inv(s, x):
        return s.a[x.cpu()].view(f"S{x.size(-1)}")[..., 0]


def dat(l, n):
    m = np.zeros((n, 26), int)
    N = np.arange(n)

    def randi():
        i = np.random.randint(0, 3, n, "B")
        return (i + 97).view("S1"), i

    def rand():
        I, i = randi()
        v = np.random.randint(0, 100, n)
        w = np.random.randint(0, 2, n, "?")
        return np.where(w, I, v.astype("S2")), np.where(w, m[N, i], v)

    def randop(O=np.array((b"&", b"|", b"^", b"+", b"-"))):
        A, a = rand()
        B, b = rand()
        o = np.random.randint(0, 5, n)
        return strcat(A, O[o], B), np.stack((a & b, a | b, a ^ b, a + b, a - b))[o, N] % 100

    q = []
    a = []
    for _ in range(l):
        I, i = randi()
        V, v = randop()
        q.append(strcat(I, b"=", V))
        a.append(np.char.rjust(v.astype("S2"), 2, b"0"))
        m[N, i] = v
    return np.stack(q), np.stack(a)
```
